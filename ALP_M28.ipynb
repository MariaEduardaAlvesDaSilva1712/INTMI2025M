3 - Crie um algoritmo que receba o nome do aluno, e 3 notas. depois Calcule a média das notas e imprima na tela:

O aluno: (aluno) está com (media) de média!

```python
# Solicita o nome do aluno
aluno = input("Digite o nome do aluno: ")

# Solicita as três notas
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

# Calcula a média
media = (nota1 + nota2 + nota3) / 3

# Exibe o resultado
print(f"O aluno {aluno} está com {media:.2f} de média!")
```

8.Peça um número de 1 a 7 e informe o dia da semana correspondente. Crie um algoritmo em Python que associe o número ao dia correto.

```python
# Solicita um número ao usuário
numero = int(input("Digite um número de 1 a 7: "))

# Cria uma lista com os dias da semana
dias_da_semana = ["Segunda-feira", "Terça-feira", "Quarta-feira",
                  "Quinta-feira", "Sexta-feira", "Sábado", "Domingo"]

# Verifica se o número está dentro do intervalo válido
if 1 <= numero <= 7:
    # Subtrai 1 porque as listas em Python começam no índice 0
    print("O dia correspondente é:", dias_da_semana[numero - 1])
else:
    print("Número inválido! Digite um número entre 1 e 7.")
```

9.Peça um ano e informe se ele é bissexto. Crie um algoritmo em Python que verifique se o ano é divisível por 4 e não por 100, exceto se também for divisível por 400.

```python
# Solicita ao usuário que digite um ano
ano = int(input("Digite um ano: "))

# Verifica se o ano é bissexto
if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    print(f"{ano} é um ano bissexto.")
else:
    print(f"{ano} não é um ano bissexto.")
```

10.Peça a quantidade de horas trabalhadas e o valor por hora. Se trabalhou até 40 horas, calcule o salário normalmente (horas × valor por hora). Se trabalhou mais de 40 horas, as horas extras devem ter um adicional de 50% (horas extras × valor por hora × 1.5). Crie um algoritmo em Python que realize esse cálculo corretamente.

```python
# Solicita a quantidade de horas trabalhadas e o valor por hora
horas_trabalhadas = float(input("Digite a quantidade de horas trabalhadas: "))
valor_hora = float(input("Digite o valor por hora: "))

# Calcula o salário
if horas_trabalhadas <= 40:
    salario = horas_trabalhadas * valor_hora
else:
    # Calcula as horas extras
    horas_extras = horas_trabalhadas - 40
    salario = (40 * valor_hora) + (horas_extras * valor_hora * 1.5)

# Mostra o resultado
print(f"O salário total é: R$ {salario:.2f}")
```

## Jogo da Adivinhação

```python
import random as rd
```

```python
aleatorio = rd.randint(0, 100)

prin("=============================================")
print("            JOGO DA ADIVINHAÇÃO             ")
print("============================================")
print("Tente adivinhar o número que estou pensando!")
print("============================================")

tentativa = 5

while (tentativa > 0):
  numero = int(input("Digite um Número: "))
  if (numero == aleatorio):
    print("Parabéns! Você acertou")
    break
  elif (numero > aleatorio):
    print("Digite um número menor:")
  else:
    print("Digite um número maior:")
  tentativa -= 1
if (tentativa == 0):
  print("")
```
