8.Peça um número de 1 a 7 e informe o dia da semana correspondente. Crie um algoritmo em Python que associe o número ao dia correto.

```python
# Solicita um número ao usuário
numero = int(input("Digite um número de 1 a 7: "))

# Cria uma lista com os dias da semana
dias_da_semana = ["Segunda-feira", "Terça-feira", "Quarta-feira",
                  "Quinta-feira", "Sexta-feira", "Sábado", "Domingo"]

# Verifica se o número está dentro do intervalo válido
if 1 <= numero <= 7:
    # Subtrai 1 porque as listas em Python começam no índice 0
    print("O dia correspondente é:", dias_da_semana[numero - 1])
else:
    print("Número inválido! Digite um número entre 1 e 7.")
```

10.Peça a quantidade de horas trabalhadas e o valor por hora. Se trabalhou até 40 horas, calcule o salário normalmente (horas × valor por hora). Se trabalhou mais de 40 horas, as horas extras devem ter um adicional de 50% (horas extras × valor por hora × 1.5). Crie um algoritmo em Python que realize esse cálculo corretamente.

```python
# Solicita a quantidade de horas trabalhadas e o valor por hora
horas_trabalhadas = float(input("Digite a quantidade de horas trabalhadas: "))
valor_hora = float(input("Digite o valor por hora: "))

# Calcula o salário
if horas_trabalhadas <= 40:
    salario = horas_trabalhadas * valor_hora
else:
    # Calcula as horas extras
    horas_extras = horas_trabalhadas - 40
    salario = (40 * valor_hora) + (horas_extras * valor_hora * 1.5)

# Mostra o resultado
print(f"O salário total é: R$ {salario:.2f}")
```

```python
import random as rd
```

```python
aleatorio = rd.randint(0, 100)

prin("=============================================")
print("            JOGO DA ADIVINHAÇÃO             ")
print("============================================")
print("Tente adivinhar o número que estou pensando!")
print("============================================")

tentativa = 5

while (tentativa > 0):
  numero = int(input("Digite um Número: "))
  if (numero == aleatorio):
    print("Parabéns! Você acertou")
    break
  elif (numero > aleatorio):
    print("Digite um número menor:")
  else:
    print("Digite um número maior:")
  tentativa -= 1
if (tentativa == 0):
  print("")
```

1. Estrutura Principal e Menu (Usando while)
O seu programa deve rodar em um loop (while) que só será interrompido quando o usuário digitar o código de saída.

A cada repetição do loop, o programa deve exibir o seguinte menu e solicitar uma opção ao usuário:

=========================
**CALCULADORA INTERATIVA**
=========================
Digite o número da opção desejada:
1 - Múltiplos de 5 em um Intervalo
2 - Contagem de Números Pares em um Intervalo
3 - Verificador de Número Primo
4 - Conversor de Temperatura (Cº para Fº)
0 - Sair do Programa
2. Tratamento das Opções (Usando if/elif/else)
Use a estrutura de controle if / elif / else para direcionar o programa para a função ou bloco de código correspondente à escolha do usuário.

Se o usuário digitar 0: A repetição deve ser interrompida (usando break) e uma mensagem de despedida deve ser exibida.

Se o usuário digitar 1, 2, 3, ou 4: O programa deve executar o cálculo solicitado (detalhado abaixo).

Se o usuário digitar qualquer outra coisa: O programa deve exibir uma mensagem de opção inválida e retornar ao menu.

3. Implementação das Opções (Usando for, while e if)
Para cada opção válida, siga as especificações abaixo. Lembre-se de sempre solicitar os dados de entrada dentro do bloco de código da opção e mostrar a resposta em seguida.

Opção	Título	Estrutura Principal Sugerida	Descrição da Tarefa
1	Múltiplos de 5	for	Peça um número inicial e um número final. Use um loop para percorrer o intervalo e imprima apenas aqueles que são múltiplos de 5.
2	Números Pares	for ou while	Peça um número inicial e um número final. Use um loop para percorrer o intervalo e contar quantos números são pares. Ao final, exiba a contagem total.
3	Número Primo	for e if	Peça um número inteiro positivo. Use um loop para verificar se o número é primo (só é divisível por 1 e por ele mesmo). Exiba se o número é ou não primo.
4	Conversor de Temperatura	Estrutura simples	Peça a temperatura em Celsius Cº. Calcule e exiba o valor convertido em Fahrenheit Fº usando a fórmula: F=Cx1.8+32
Após a execução de qualquer cálculo (opções 1 a 4), o programa deve retornar automaticamente ao menu principal para que o usuário possa fazer outra escolha.

```python
# ========================================
#      CALCULADORA INTERATIVA EM PYTHON
# ========================================

while True:
    print("=========================")
    print("   CALCULADORA INTERATIVA")
    print("=========================")
    print("Digite o número da opção desejada:")
    print("1 - Múltiplos de 5 em um Intervalo")
    print("2 - Contagem de Números Pares em um Intervalo")
    print("3 - Verificador de Número Primo")
    print("4 - Conversor de Temperatura (Cº para Fº)")
    print("0 - Sair do Programa")

    opcao = input("Escolha uma opção: ")

    # ------------------- OPÇÃO 0: SAIR -------------------
    if opcao == "0":
        print("Encerrando o programa... Até mais!")
        break

    # ------------------- OPÇÃO 1 -------------------------
    elif opcao == "1":
        print("\n--- MÚLTIPLOS DE 5 ---")
        inicio = int(input("Digite o número inicial: "))
        fim = int(input("Digite o número final: "))

        print(f"\nMúltiplos de 5 entre {inicio} e {fim}:")
        for num in range(inicio, fim + 1):
            if num % 5 == 0:
                print(num)

        print()  # Linha vazia para organização

    # ------------------- OPÇÃO 2 -------------------------
    elif opcao == "2":
        print("\n--- CONTAGEM DE NÚMEROS PARES ---")
        inicio = int(input("Digite o número inicial: "))
        fim = int(input("Digite o número final: "))

        cont = 0
        for num in range(inicio, fim + 1):
            if num % 2 == 0:
                cont += 1

        print(f"\nQuantidade de números pares no intervalo: {cont}\n")

    # ------------------- OPÇÃO 3 -------------------------
    elif opcao == "3":
        print("\n--- VERIFICADOR DE NÚMERO PRIMO ---")
        n = int(input("Digite um número inteiro positivo: "))

        if n <= 1:
            print(f"{n} NÃO é primo (primos começam em 2).\n")
        else:
            eh_primo = True
            for i in range(2, n):
                if n % i == 0:
                    eh_primo = False
                    break

            if eh_primo:
                print(f"{n} é PRIMO!\n")
            else:
                print(f"{n} NÃO é primo.\n")

    # ------------------- OPÇÃO 4 -------------------------
    elif opcao == "4":
        print("\n--- CONVERSOR DE CELSIUS PARA FAHRENHEIT ---")
        c = float(input("Digite a temperatura em Celsius: "))
        f = c * 1.8 + 32

        print(f"\n{c}ºC equivalem a {f:.2f}ºF\n")

    # ------------------- OPÇÃO INVÁLIDA -------------------
    else:
        print("\nOpção inválida! Tente novamente.\n")
```
